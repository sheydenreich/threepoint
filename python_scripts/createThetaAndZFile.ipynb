```python
import numpy as np
```

```python
thetas=np.array([4,8,16,32])
zs=np.array([2,3,4,5]) #np.array([0,1,2,3,4])
```

```python
thetas_list=[]
zs_list=[]

for i in range(len(zs)):
    z1=zs[i]
    for j in range(i, len(zs)):
        z2=zs[j]
        for k in range(j, len(zs)):
            z3=zs[k]
            zs_tmp=np.array([z1, z2, z3])
            for l in range(len(thetas)):
                th1=thetas[l]
                for m in range(l, len(thetas)):
                    th2=thetas[m]
                    for n in range(m, len(thetas)):
                        th3=thetas[n]
                        ths_tmp=np.array([th1, th2, th3])
                        zs_list.append(zs_tmp)
                        thetas_list.append(ths_tmp)
len(thetas_list)
```

```python
thetas_list_onlyEqui=[]
zs_list_onlyEqui=[]

for i in range(len(zs)):
    z1=zs[i]
    for j in range(i, len(zs)):
        z2=zs[j]
        for k in range(j, len(zs)):
            z3=zs[k]
            zs_tmp=np.array([z1, z2, z3])
            for l in range(len(thetas)):
                th1=thetas[l]
                ths_tmp=np.array([th1, th1, th1])
                zs_list_onlyEqui.append(zs_tmp)
                thetas_list_onlyEqui.append(ths_tmp)
len(thetas_list_onlyEqui)
```
